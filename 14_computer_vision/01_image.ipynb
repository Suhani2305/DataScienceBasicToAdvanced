```python
import cv2 as cv
import warnings
warnings.filterwarnings('ignore')
import numpy as np

# read a image
imag = cv.imread('images\image.png')
```

```python
# for showing image
cv.imshow('Image', imag)
cv.waitKey(10000)  # wait for key press to close window
cv.destroyAllWindows()   # close all windows
# cv.destroyWindow('Image')  # close specific window
```

```python
#  resize image 
image = cv.resize(imag, (400, 400))
cv.imshow('Image1', image)
cv.waitKey(0)
cv.destroyAllWindows()
```
