```python
import numpy as np
import cv2 as cv
# import webcame using cv inside cv
cap = cv.VideoCapture(0)
if cap.isOpened():
    ret, frame = cap.read()
else:
    ret = False
    
while ret:
    ret, frame = cap.read()
    cv.imshow('Webcam', frame)
    if cv.waitKey(25) & 0xFF == ord('q'):  # ESC key to break
        break
    
cap.release()
cv.destroyAllWindows()
```

```python
import numpy as np
import cv2 as cv

# Open webcam
cap = cv.VideoCapture(0)

if not cap.isOpened():
    print("❌ Webcam not accessible")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # 🔹 Convert BGR → Grayscale
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)

    # Show grayscale webcam
    cv.imshow('Webcam Grayscale', gray)

    if cv.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```
