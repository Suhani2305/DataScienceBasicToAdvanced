```python
print(arr.ndim)  # number of dimensions
print(arr.shape) # shape of the array
print(arr.size)  # total number of elements
print(arr.dtype) # data type of elements
```

```python
# 2d numpy array
arr2d = np.array([[1,2,3],[4,5,6]])
print(arr2d)
print(arr2d.ndim)
print(arr2d.shape)
print(arr2d.size)
print(arr2d.dtype)
```

```python
# make array using zeros and ones
arr_zeros = np.zeros((3,4))
print(arr_zeros)
print(arr_zeros.dtype)
arr_ones = np.ones((2,3))
print(arr_ones)

# chnange data type
arr_3 = np.zeros((2,3), dtype = 'int32')
print(arr_3)
```

```python
# arange , linespace, logspace, eye, identity, diag, random, empty, full
arr_arange = np.arange(0,10,2)
print(arr_arange)
arr_linspace = np.linspace(0,13,5, dtype = 'int32')
print(arr_linspace)
arr_logspace = np.logspace(0,3,4)
print(arr_logspace)
arr_eye = np.eye(3)
print(arr_eye)
arr_identity = np.identity(4)
print(arr_identity)
arr_diag = np.diag([1,2,3,4])
print(arr_diag)
arr_random = np.random.rand(3,4)
print(arr_random)
arr_empty = np.empty((2,3))
print(arr_empty)
arr_full = np.full((2,3),7)
print(arr_full)
```

```python
c = np.array([0,2,0,1,0,0,5,3,7])
print(np.nonzero(c))  # index of non zero elements
print(c.astype('bool'))  # change to boolean array
```

```python
# difference between copy and view : copy creates a new array and view creates a view of the original array and if we change the view the original array will also change but if we change the copy the original array will not change

c = arr.copy()
d = arr.view()
print(c)
print(d)

# chnge in view  and copy
d[0] = 100
print(arr)
print(d)
c[3] = 200
print(arr)
print(c)
```
