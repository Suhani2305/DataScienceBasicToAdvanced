```python
# import lab
import numpy as np
import pandas as pd
df = pd.read_csv('data.csv')
df.head()
```

```python
# handling missing value
df.isnull().sum()
```

```python
columns = df.columns
for col in columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype == 'object':
            df[col].fillna(df[col].mode()[0], inplace=True)
        else:
            df[col].fillna(df[col].mean(), inplace=True)
```
