```python
# fibonacci series
n = int(input("Enter number of terms: "))
a, b = 0, 1
count = 0
if n <= 0:
    print("Please enter a positive integer")
elif n == 1:
    print("Fibonacci sequence upto", n, ":")
    print(a)
else:
    print("Fibonacci sequence:")
    while count < n:
        print(a)
        a, b = b, a + b
        count += 1
        
```

```python
# factorial of a number
num = int(input("Enter a number: "))
factorial = 1
if num < 0:
    print("Sorry, factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print("The factorial of", num, "is", factorial)
```

## Mean , Median and Mode

```python
# mean median and mode
import pandas as pd
import numpy as np
import scipy.stats as stats

data = [12,43,56,90,3,98,56,70,31,56,34,34,34]
print("Mean:", np.mean(data))
print("Median:", np.median(data))

print("Mode:", pd.Series(data).mode()[0])

print("Mode using scipy:", stats.mode(data))
```

```python
# standard deviation and variance
print("Standard Deviation:", np.std(data) )
print("Variance:", np.var(data))
```

```python
# percentile and quartile
print("25th Percentile:", np.percentile(data, 25))
print("50th Percentile:", np.percentile(data, 50))
print("75th Percentile:", np.percentile(data, 75))
print("1st Quartile (Q1):", np.percentile(data, 25))
print("2nd Quartile (Q2):", np.percentile(data, 50))
print("3rd Quartile (Q3):", np.percentile(data, 75))
```

```python
# create a list of 250 random flot humber betwwen 0 to 5
import numpy as np
np.random.seed(0)
random_floats = np.random.normal(5,1, 100000)
print(random_floats)
```
